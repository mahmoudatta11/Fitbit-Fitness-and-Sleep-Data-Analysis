```python
df1['ActivityDate'] = pd.to_datetime(df1['ActivityDate'])
df1.info()
```

```python
df2['SleepDay'] = pd.to_datetime(df2['SleepDay'])
df2.info()
```

```python
df1.rename(columns={'ActivityDate': 'Date'}, inplace=True)
df2.rename(columns={'SleepDay': 'Date'}, inplace=True)
```

```python
df = pd.merge(df1, df2, on=['Id', 'Date'], how='left')
df
```

```python
df.info()
```

```python
df.dtypes
```

```python
avg_sleeprecord = df['TotalSleepRecords'].astype(float).mean()
df['TotalSleepRecords']= df['TotalSleepRecords'].replace(np.nan, avg_sleeprecord)

avg_minasleep = df['TotalMinutesAsleep'].astype(float).mean()
df['TotalMinutesAsleep']= df['TotalMinutesAsleep'].replace(np.nan, avg_minasleep)

avg_timeinbed = df['TotalTimeInBed'].astype(float).mean()
df['TotalTimeInBed']= df['TotalTimeInBed'].replace(np.nan, avg_timeinbed)
df
```

```python
sns.set(style='whitegrid')
plt.figure(figsize=(10, 5))

sns.regplot(
    x='TotalSteps', y='Calories', data=df,
    scatter_kws={'alpha':0.4},  # makes overlapping dots translucent
    line_kws={'color':'#FF0800','alpha':0.6})  # adds a regression line
plt.title('Total Steps vs Calories Burned')
plt.xlabel('Total Steps')
plt.ylabel('Calories Burned')
plt.legend(['Calories Burned'])
plt.show()
```

```python
import matplotlib.pyplot as plt

# 1. Calculate average minutes for each activity category
categories = ['Sedentary', 'Lightly Active', 'Fairly Active', 'Very Active']
minutes = [
    df['SedentaryMinutes'].mean(),
    df['LightlyActiveMinutes'].mean(),
    df['FairlyActiveMinutes'].mean(),
    df['VeryActiveMinutes'].mean()
]

# 2. Modern, professional color palette (Dark Slate, Teal, Coral, Soft Gold)
colors = ['#2D3748', '#319795', '#ED8936', '#ECC94B']

fig, ax = plt.subplots(figsize=(6, 4), subplot_kw=dict(aspect="equal"), dpi=150)

# 3. Create Donut Slices (labels set to None to keep slices clean)
wedges, texts, autotexts = ax.pie(
    minutes,
    autopct='%1.1f%%',
    pctdistance=0.80,         # Position percentages neatly inside slices
    startangle=140,
    colors=colors,
    wedgeprops=dict(width=0.35, edgecolor='white', linewidth=2.5) # Sleek ring thickness
)

# 4. Format the percentage labels inside the slices
plt.setp(autotexts, size=9, weight="bold", color="white")

# 5. Add a clean Legend outside the chart anchored on the right
ax.legend(
    wedges, 
    categories,
    title="Activity Intensity",
    loc="center left",
    bbox_to_anchor=(1, 0.5),   # Moves legend cleanly outside the plot area
    frameon=False,             # Removes clunky border around legend box
    fontsize=11,
    title_fontsize=12
)

# 6. Add summary metric right inside the center hole
total_hours = sum(minutes) / 60
ax.text(
    0, 0, 
    f"Daily Average\n{total_hours:.1f} Hours", 
    ha='center', 
    va='center', 
    fontsize=12, 
    fontweight='bold', 
    color='#2D3748'
)

# 7. Title formatting
plt.title("Daily Active vs. Sedentary Time Distribution", fontsize=14, fontweight='bold', pad=20, loc='left')

plt.tight_layout()
plt.show()
```

```python
df['DayOfWeek'] = df['Date'].dt.day_name()
days_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

plt.figure(figsize=(9, 5))
sns.barplot(
        data=df, 
        x='DayOfWeek', 
    y='TotalSteps', 
    order=days_order, 
    ci=None, 
    palette='Blues_d'
)
plt.axhline(10000, color='red', linestyle='--', label='10,000 Step Goal')
plt.title('Average Steps Taken by Day of the Week')
plt.xlabel('Day of Week')
plt.ylabel('Average Total Steps')
plt.legend()
plt.show()
```

```python
# Count active days per user
usage_per_user = df.groupby('Id')['Date'].nunique().reset_index()
usage_per_user.columns = ['Id', 'DaysTracked']
usage_per_user
```

```python
# Categorize usage
def usage_category(days):
    if days >= 21:
        return 'High Usage (21-31 Days)'
    elif days >= 11:
        return 'Moderate Usage (11-20 Days)'
    else:
        return 'Low Usage (1-10 Days)'

usage_per_user['UsageCategory'] = usage_per_user['DaysTracked'].apply(usage_category)

plt.figure(figsize=(8, 5))
sns.countplot(
    data=usage_per_user, 
    x='UsageCategory', 
    order=['High Usage (21-31 Days)', 'Moderate Usage (11-20 Days)', 'Low Usage (1-10 Days)'],
    palette='Set2'
)
plt.title('User Classification by Wear Frequency')
plt.xlabel('Usage Category')
plt.ylabel('Number of Users')
plt.show()
```
